## Los 4 errores más comunes de los estudiantes

1. **No usar comillas para strings**  
   Se asignan textos sin comillas (ej. `estado_civil = casado`), Python espera una variable y falla.

2. **Usar “sí/no” en lugar de booleanos**  
   Se escribe `si`/`no` en vez de `True`/`False`, que son los booleanos válidos en Python.

3. **Nombres de variables inválidos**  
   Se usan espacios o acentos (ej. `color pantalón`). Use letras, números y guion bajo (`color_pantalon`).

4. **Errores al manipular strings e índices**  
   Uso incorrecto de `.upper()`, `.strip()`, `.isalpha()` u olvidar que los índices empiezan en 0.

    <strong>Ayuda para resolver el problema:</strong>
    <ul style="margin:8px 0 0 18px;">
      <li>Elimina espacios con <code>.replace(" ", "")</code>.</li>
      <li>Usa <code>.startswith("ATG")</code> para verificar el inicio.</li>
      <li>Cuenta con <code>.count("CCT")</code>.</li>
      <li>Encuentra la posición con <code>.find("ATT")</code>.</li>
      <li>Convierte a minúsculas con <code>.lower()</code>.</li>
      <li>Reemplaza <code>t</code> por <code>u</code> con <code>.replace("t", "u")</code>.</li>
    </ul>
  </div>
</div>


````python
cadena = "ATG GCC TTA CCT ATT GGA CCT TGA"

# 1 Eliminar los espacios en blanco de la cadena de ADN.

sin_espacios = cadena.replace(" ","")   # - .replace( "TEXTO EN EL STRING" , "TEXTO NUEVO")

print(sin_espacios)

# 2 Comprobar si la secuencia inicia con el nucleótido ATG.
empieza_ATG = sin_espacios.startswith("ATG")  # - STARTSWITH("TEXTO") - BOOLEANO TRUE/FALSE ¿La primera paalbra es TEXTO?
print(f" Empieza con ATG? {empieza_ATG}")

# 3 - Contar cuántas veces aparece el patrón CCT.

num_apariciones = sin_espacios.count("CCT") # - .count("TEXTO") VECES QUE APARECE EL TEXTO

print(num_apariciones)

# 4 - Indicar el índice donde se encuentra por primera vez el nucleótido ATT.

indice_ATT = sin_espacios.find("ATT") # .find("TEXTO") PRIMERA APARICION DEL TEXTO ME EL INDICE

print(indice_ATT)

# 5 - Convertir toda la secuencia a minúsculas.

secuencia_minusculas = sin_espacios.lower()   # .lower() VOLVER A MINUSCULAS TODO EL STRING!!! - CONTRARIO A .upper()

print(secuencia_minusculas)

# 6 - Reemplazar la letra t por u para obtener la versión de ARN.

cadena_reemplazo = secuencia_minusculas.replace("t","u")

print(cadena_reemplazo)


# 7 - Imprimir los resultados

print("La cadena resultantes, tras aplicar minusculas, eliminar espacios, y reemplazar la t por la u, es:" , cadena_reemplazo)
````
