<h2>1) Uso de <code>.sort_values()</code> para ordenar DataFrames por una o varias columnas. Parámetro <code>inplace=True</code></h2>

<p>
El método <code>.sort_values()</code> ordena las filas de un DataFrame según los valores de una o más columnas.
</p>

<h3>1.1 Ordenar por una columna</h3>
<pre><code>import pandas as pd

df = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Sofía", "Pedro"],
    "edad": [21, 19, 21, 18],
    "nota": [5.8, 6.4, 5.9, 5.1]
})

df_ordenado = df.sort_values(by="nota")
df_ordenado</code></pre>

<h3>1.2 Ordenar por varias columnas</h3>
<p>
Cuando se ordena por varias columnas, primero se usa la primera columna indicada y, si hay empates, se utiliza la siguiente.
</p>
<pre><code>df_ordenado2 = df.sort_values(by=["edad", "nota"])
df_ordenado2</code></pre>

<h3>1.3 Parámetro <code>inplace=True</code></h3>
<p>
Si se usa <code>inplace=True</code>, el DataFrame original se modifica directamente.
En este caso, el método no entrega un nuevo DataFrame para asignar.
</p>
<pre><code>df.sort_values(by="nota", inplace=True)
df</code></pre>

<p>
Evite asignar el resultado cuando use <code>inplace=True</code>.
</p>
<pre><code># Ejemplo incorrecto (evitar):
# df = df.sort_values(by="nota", inplace=True)</code></pre>

<h2>2) Diferencia entre datos categóricos y no categóricos</h2>

<h3>2.1 Datos categóricos</h3>
<p>
Son datos que representan categorías o grupos. Sirven para clasificar.
</p>
<ul>
  <li>Ejemplos: sección, carrera, ciudad, tipo, estado.</li>
  <li>Se suelen analizar con conteos y frecuencias.</li>
</ul>

<h3>2.2 Datos no categóricos</h3>
<p>
Son datos que representan cantidades o mediciones, normalmente numéricas.
</p>
<ul>
  <li>Ejemplos: edad, nota, salario, tiempo.</li>
  <li>Se suelen analizar con operaciones como promedio, mínimo o máximo.</li>
</ul>

<p>
La diferencia es importante porque los análisis cambian según el tipo de dato.
Por ejemplo, es útil contar cuántas veces aparece una categoría, pero no tiene sentido promediar nombres de categorías.
</p>

<p>
Estas funciones permiten explorar valores categóricos, contar frecuencias y filtrar por pertenencia.

<p>
Cuenta cuántas veces aparece cada valor en una columna.

<pre><code>df2["seccion"].value_counts()</code></pre>

<h3>3.3 <code>isin()</code></h3>
<p>
Indica si cada valor de una columna pertenece a un conjunto de valores.
El resultado es una serie booleana que se puede usar para filtrar.
</p>
<pre><code>mask = df2["seccion"].isin(["A", "C"])
mask</code></pre>

<p>
Filtrado usando <code>isin()</code>:
</p>
<pre><code>df2_filtrado = df2[df2["seccion"].isin(["A", "C"])]

````python
# Solución Ejercicio 1: Ordenamiento y tipo de datos
# -----------------------------------------------

import pandas as pd

# DataFrame base
df = pd.DataFrame({
    "estudiante": ["Ana", "Luis", "Sofía", "Pedro", "Camila"],
    "seccion": ["A", "B", "A", "C", "B"],
    "edad": [21, 19, 21, 18, 20],
    "nota": [5.8, 6.4, 5.9, 5.1, 6.0]
})

# 1) Ordenar por la columna "nota" y guardar en una nueva variable
# No se usa inplace, por lo que se genera un nuevo DataFrame
df_nota = df.sort_values(by="nota")

# 2) Ordenar por "seccion" y luego por "edad"
# Primero se ordena por la sección y, en caso de empate, por la edad
df_seccion_edad = df.sort_values(by=["seccion", "edad"])

# 3) Ordenar el DataFrame original por "edad" usando inplace=True
# El DataFrame df se modifica directamente
df.sort_values(by="edad", inplace=True)

# 4) Identificación de columnas categóricas y no categóricas
# Categóricas: representan etiquetas o grupos
categoricas = ["estudiante", "seccion"]

# No categóricas: representan mediciones numéricas
no_categoricas = ["edad", "nota"]

# Mostrar resultados
print(df_nota)
print(df_seccion_edad)
print(df)
print(categoricas)
print(no_categoricas)
````

<pre><code>import pandas as pd

df = pd.DataFrame({
    "estudiante": ["Ana", "Luis", "Sofía", "Pedro", "Camila", "Jorge"],
    "seccion": ["A", "B", "A", "C", "B", "A"],
    "nota": [5.8, 6.4, 5.9, 5.1, 6.0, 5.5]
})

df</code></pre>

<ol>
  <li>
    Obtén los valores únicos de la columna <code>seccion</code>.
  </li>
  <li>
    Cuenta cuántas veces aparece cada sección usando <code>value_counts()</code>.
  </li>
  <li>

````python
# Solución Ejercicio 2: Exploración de datos categóricos
# ----------------------------------------------------

import pandas as pd

# DataFrame base
df = pd.DataFrame({
    "estudiante": ["Ana", "Luis", "Sofía", "Pedro", "Camila", "Jorge"],
    "seccion": ["A", "B", "A", "C", "B", "A"],
    "nota": [5.8, 6.4, 5.9, 5.1, 6.0, 5.5]
})

# 1) Valores únicos de la columna categórica "seccion"
# Permite identificar qué categorías existen
secciones_unicas = df["seccion"].unique()

# 2) Conteo de frecuencia de cada sección
# Muestra cuántos estudiantes hay por sección
conteo_secciones = df["seccion"].value_counts()

# 3) Filtrado usando isin()
# Se seleccionan solo las filas cuya sección sea A o B
df_filtrado = df[df["seccion"].isin(["A", "B"])]

# Mostrar resultados
secciones_unicas
conteo_secciones
df_filtrado
````

````python
# Solución Ejercicio 3: Promedio por categoría con groupby
# ------------------------------------------------------

import pandas as pd

# DataFrame base
df = pd.DataFrame({
    "estudiante": ["Ana", "Luis", "Sofía", "Pedro", "Camila", "Jorge"],
    "seccion": ["A", "B", "A", "C", "B", "A"],
    "nota": [5.8, 6.4, 5.9, 5.1, 6.0, 5.5]
})

# Agrupar por la columna categórica "seccion"
# y calcular el promedio de la columna "nota"
promedio_por_seccion = df.groupby("seccion")["nota"].mean()

# Mostrar resultado
promedio_por_seccion
````
