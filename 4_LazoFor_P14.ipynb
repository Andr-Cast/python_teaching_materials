<h2>Ejercicio 2.6 — Verificación de números triangulares</h2>

<p><b>Enunciado:</b></p>

<p>
Cree una función llamada <code>es_triangular(numero)</code> que determine si un número pertenece a la sucesión de números triangulares.
</p>

<ol>
  <li>Inicie con una lista que contenga el valor inicial <code>[1]</code>.</li>
  <li>Utilice un lazo <code>for</code> con un rango de <code>numero + 10</code> para generar más números triangulares.</li>
  <li>En cada iteración, sume el número siguiente al último valor de la lista y agréguelo con <code>append()</code>.</li>
  <li>Devuelva <code>True</code> o <code>False</code> dependiendo de si el número está dentro de la lista.</li>
</ol>

<p>Luego, en el programa principal:</p>

<ol>
  <li>Solicite al usuario ingresar 10 números, uno por uno.</li>
  <li>Verifique si la entrada es un número entero:
    <ul>
      <li>Si lo es, agréguelo a una lista.</li>
      <li>Si no lo es, muestre <code>"La entrada es inválida."</code></li>
    </ul>
  </li>
  <li>Recorra la lista validando cada número:
    <ul>
      <li>Si el número está fuera del rango <code>[5, 200]</code>, muestre <code>"No rango"</code>.</li>
      <li>Si está dentro, llame a la función <code>es_triangular()</code> para verificarlo.</li>
    </ul>
  </li>
  <li>Muestre los resultados en una tabla con el siguiente formato:</li>
</ol>

<pre>
| Número | Triangular |
|--------|-------------|
|      3 | Sí          |
|     10 | Sí          |
|     11 | No          |
|    250 | No rango    |
</pre>


```python
def es_triangular(numero):
  triangulares = [1]

  for num in range( 2 , numero + 10):
    nuevo = triangulares[-1] + num

    triangulares.append(nuevo)

  return numero in triangulares


entradas = []

for _ in range(10):
  entrada = input("Ingrese un numero:")

  if entrada.isdigit():
    entradas.append(int(entrada))
  else:
    print("La entrada no es un numero, es invalida")



print("| Numero | Triangular |")
print("|--------+-------------|")

for numero in entradas:
  if numero < 5 or numero > 200:
    print(f"|{numero:6d} | {"No Rango":<9}|")
  else:
    if es_triangular(numero):
      print(f"|{numero:6d} | {"Sí":<9}|")
    else:
      print(f"|{numero:6d} | {"No":<9}|")
```

<h2>Ejercicio 2.7 — Cálculo de múltiplos usando bucle For</h2>

<p><b>Enunciado:</b></p>

<p>
Cree una función llamada <code>listar_multiplos(numero, limite)</code> que genere y devuelva una lista con todos los múltiplos de un número dentro del rango indicado.
</p>

<ol>
  <li>Defina la función <code>listar_multiplos(numero, limite)</code> con dos parámetros:
    <ul>
      <li><code>numero</code>: el número base cuyos múltiplos se calcularán.</li>
      <li><code>limite</code>: el valor máximo que no debe superarse.</li>
    </ul>
  </li>
  <li>Dentro de la función:
    <ul>
      <li>Cree una lista vacía llamada <code>multiplos</code>.</li>
      <li>Utilice un lazo <code>for</code> con <code>range()</code> para recorrer los números desde <code>numero</code> hasta <code>limite</code>.</li>
      <li>En cada iteración, verifique si el número actual es múltiplo del valor base usando <code>%</code>.</li>
      <li>Si cumple la condición, agréguelo a la lista.</li>
      <li>Devuelva la lista al final de la función.</li>
    </ul>
  </li>
</ol>

<p>Luego, en el programa principal:</p>

<ol>
  <li>Solicite al usuario ingresar un número base y un límite superior.</li>
  <li>Valide que ambas entradas sean números enteros; si alguna no lo es, muestre <code>"Entrada inválida"</code>.</li>
  <li>Llamar a la función <code>listar_multiplos()</code> y guardar el resultado.</li>
  <li>Muestre los múltiplos en una tabla con el formato siguiente:</li>
</ol>

<pre>
| Índice | Múltiplo |
|--------|-----------|
|      1 | 5         |
|      2 | 10        |
|      3 | 15        |
|      4 | 20        |
</pre>


```python
def listar_multiplos(numero, limite):
  multipos = []

  for num in range (numero, limite + 1):

    # Verificar que "num" es multiplo del parametro "numero"

    if num % numero == 0:
      multipos.append(num)

  return multipos


n_base = input("Ingrese un numero base:")
limite = input("Ingrese un limite:")

if not (n_base.isdigit() and limite.isdigit()):

  print("Entrada invalida")

else:

  lista = listar_multiplos(int(n_base), int(limite))

  print("| Indice | Multiplo |")
  print("|--------+----------|")

  for indice in range(len(lista)):

    print(f"|{indice+1:7d} | {lista[indice]:9d}|")








```

<h2>Ejercicio 2.8 — Análisis de texto con bucles For</h2>

<p><b>Enunciado:</b></p>

<p>
Cree un programa que analice un texto ingresado por el usuario y calcule distintas estadísticas usando el bucle <code>for</code>.
</p>

<ol>
  <li>Solicite al usuario ingresar una oración o párrafo corto.</li>
  <li>Recorra el texto con un lazo <code>for</code> y calcule los siguientes valores:
    <ul>
      <li>Cantidad total de caracteres.</li>
      <li>Cantidad de vocales (<code>a, e, i, o, u</code>).</li>
      <li>Cantidad de consonantes (solo letras del alfabeto).</li>
      <li>Cantidad de espacios.</li>
      <li>Cantidad de signos de puntuación (<code>.,;:!?¡¿</code>).</li>
    </ul>
  </li>
  <li>Utilice variables contadoras para cada tipo de elemento.</li>
  <li>Ignore mayúsculas o minúsculas al contar letras.</li>
  <li>Al finalizar el recorrido, muestre un resumen con los resultados de cada contador, por ejemplo:</li>
</ol>

<pre>
Texto analizado: "Hola mundo, ¿cómo estás?"
Total de caracteres: 25
Vocales: 8
Consonantes: 10
Espacios: 3
Signos de puntuación: 4
</pre>

<ol start="6">
  <li>Evite usar funciones integradas como <code>count()</code> o librerías externas.</li>
  <li>Todo el análisis debe realizarse únicamente con estructuras de control y bucles <code>for</code>.</li>
</ol>


```python
texto = input("Ingrese su texto:")

total_caracteres = 0
vocales = 0
consonantes = 0
espacios = 0
signos_de_puntuacion = 0

lista_vocales = ["a","e","i","o","u"]
lista_signos = ".,;:!?¿"

for caracter in texto:
  total_caracteres += 1

  if caracter == " ":
    espacios += 1

  elif caracter in lista_signos:
    signos_de_puntuacion += 1

  elif caracter.lower() in lista_vocales:
    vocales += 1

  elif caracter.isalpha():

    consonantes += 1
  else:
    print("No esta en lo esperado")


print(f"Texto analizado {texto}")
print(f"Total de caracteres: {total_caracteres}")
print(f"Vocales: {vocales}")
print(f"Consonantes: {consonantes}")
print(f"Espacios: {espacios}")
print(f"Signos de puntuacion: {signos_de_puntuacion}")

```

<h2>Ejercicio 2.10 — Secuencia de Collatz</h2>

<p>
La <b>secuencia de Collatz</b> se define para un número entero mayor que 1 de la siguiente manera:
</p>

<ul>

  <li>Si el número es impar, se multiplica por 3 y se suma 1.</li>

<p>

</p>

<pre>
