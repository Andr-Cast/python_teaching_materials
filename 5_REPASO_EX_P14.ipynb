<h1>Tema 1 - Función obtener_temperatura(valor)</h1>

<h2>Enunciado</h2>
<p>
Crea una función llamada <code>obtener_temperatura(valor)</code> que reciba una cadena de caracteres llamada <code>valor</code>.
La función debe devolver el número flotante correspondiente a una temperatura en grados Celsius
solo si la cadena tiene un formato numérico válido seguido del carácter <code>C</code> o <code>c</code>.
En caso contrario, la función debe devolver <code>0.0</code>.
</p>

<h2>Reglas de formato</h2>
<ul>
  <li>El punto <code>.</code> es el único carácter especial permitido en la parte numérica y actúa como separador decimal.</li>
  <li>El carácter <code>C</code> o <code>c</code> debe aparecer únicamente al final de la cadena.</li>
  <li>No puede haber más de un punto decimal en la parte numérica.</li>
  <li>La parte numérica debe contener al menos un dígito.</li>
</ul>

<h2>Ejemplos</h2>
<pre>
"23.5C"    devuelve 23.5
"0c"       devuelve 0.0
"45.0cC"   devuelve 0.0
"7.5F"     devuelve 0.0
"27..5c"   devuelve 0.0
"12C"      devuelve 12.0
".5c"      devuelve 0.5
</pre>


```python
def obtener_temperatura(valor):
  # VALOR GENERICO 25.5C# NO TIENE EL FORMATO BUSCADO

  # La longitud del valor, debe de ser MAYOR A 1, para que haya almenos 1 digito, y contando la LETRA C
  # La longitud final, debe de ser mayor a 2

  if len(valor) < 2:
    print("Hola")
    return 0.0

  caracter_final = valor[-1].lower()   # Este caracter, debe de ser "c" minuscula, ese es el formato

  if caracter_final != "c":
    # No se cumple el formato
    return 0.0


  parte_numerica = valor[:-1] # Solo tomamos en cuenta los numeros, no la letra "C"

  for caracter in parte_numerica:
    if not caracter.isdigit() and caracter != ".":
      return 0.0

  # Contar la cantidad de puntos decimales, "."

  cantidad_puntos = 0
  for caracter in parte_numerica:
    if caracter == ".":
      cantidad_puntos += 1

  if cantidad_puntos > 1:
    return 0.0

  return float(parte_numerica)

a = obtener_temperatura("1")
print(a)

```

<h1>Tema 2 - Función monto_categoria(lista_proyectos, categoria)</h1>

<h2>Enunciado</h2>
<p>
Implementa la función <code>monto_categoria(lista_proyectos, categoria)</code> para calcular el monto total
de los proyectos asociados a una categoría específica.
</p>

<h2>Formato de los datos</h2>
<p>
Cada elemento de la lista <code>lista_proyectos</code> tiene el siguiente formato:
</p>
<pre>
Categoría;proyecto1:valor;proyecto2:valor;proyecto3:valor;...
</pre>

<h2>Ejemplo de lista de proyectos por categoría</h2>
<pre>
lista_proyectos = [
  "Ciencias Naturales;Monitor cardíaco:230;Primeros auxilios:210",
  "Tecnología;Clasificador de imágenes:250;Chatbot para Call Center:220",
  "Arte y Creatividad;App para matemáticas:180;Juego de lógica para niños:160"
]
</pre>

<h2>La función debe:</h2>
<ul>
  <li>Buscar la categoría especificada <strong>sin distinguir mayúsculas/minúsculas</strong>.</li>
  <li>Sumar los montos de todos los proyectos de esa categoría.</li>
  <li>Retornar el total calculado.</li>
</ul>

<h2>Ejemplo de uso</h2>
<pre>
cat = "Ciencias naturaleS"
resultado = monto_categoria(lista_proyectos, cat)
# Salida esperada: 440.00
</pre>

```python
def monto_categoria(lista_proyectos, categoria):
  monto_total = 0.0
  categoria_buscada = categoria.lower()

  for elemento in lista_proyectos:
    partes = elemento.split(";") # -- Ciencias naturales, Monitorcardia:230 , Primeros Auxilios:210, Proyecto3:valor3....
                                                     #    [Monitorcardia , 230]
                                                    #           0           1
    categoria_actual = partes[0].lower()

    if categoria_actual == categoria_buscada:

      for i in range(1, len(partes)):
        valor = partes[i].split(":")[1]
        numero = float(valor)
        monto_total += numero

  return monto_total




```

<h1>Tema 3 - Carrera entre Liebre y Tortuga</h1>

<h2>Enunciado</h2>
<p>
Implementa un programa que simule una carrera de <strong>10 turnos</strong> entre dos animales: <strong>"Liebre"</strong> y <strong>"Tortuga"</strong>.
Cada animal tiene una velocidad máxima almacenada en las variables <code>vel_l</code> y <code>vel_t</code>, respectivamente.
Cada turno, cada animal avanza una cantidad aleatoria de casillas entre <strong>1</strong> y su velocidad máxima (ambos incluidos).
</p>

<p>
Existe una lista llamada <code>l_trampas</code> que indica las casillas donde hay trampas para la liebre.
Si la liebre cae en una trampa, debe regresar a la posición anterior (donde estaba antes de ese turno).
</p>

<h2>Reglas del avance</h2>
<ul>
  <li>El avance de cada animal se acumula turno a turno.</li>
  <li>Las casillas recorridas en cada movimiento se suman a su posición total.</li>
  <li>Si la liebre cae en una trampa, retrocede a su posición anterior.</li>
  <li>En cada turno se debe mostrar con guiones el avance de cada animal.</li>
  <li>Al finalizar los 10 turnos:
    <ul>
      <li>Gana el animal con mayor cantidad de casillas recorridas.</li>
      <li>Si ambos tienen la misma cantidad, se declara <strong>empate</strong>.</li>
    </ul>
  </li>
</ul>

<h2>Datos iniciales</h2>
<pre>
vel_l = 5  # velocidad máxima de la liebre
vel_t = 3  # velocidad máxima de la tortuga
l_trampas = [4, 7, 9, 13]
</pre>

<h2>Ejemplo de salida esperada</h2>
<pre>
Turno 1:
Liebre: --- (posición 3)
Tortuga: -- (posición 2)

Turno 2:
Liebre cayó en trampa en casilla 4, regresa a su posición anterior
Liebre: --- (posición 3)
Tortuga: ----- (posición 5)

... otros turnos ...

Turno 10:
Liebre: --------- (posición 9)
Tortuga: -------------- (posición 14)

Ganador: Tortuga
</pre>


```python
import random


# HAY QUE CONTAR LAS POSICIONES --- CONTADOR - ACUMULADOR
# HACER ALGO N VECES, 10 TURNOS --- BUCLE FOR
# NO TENGO IDEA QUE HAY QUE HACER -- DEJA ESPACIO EN BLANCO X SI ACASO


# IDENTACIONES - PUNTUACION (DOS PUNTOS, +=....) - RANGOS DE INDEXACION - MAYUSCULAS MINUSCULAS


# La identacion se utiliza en:  FOR - DEF - IF/ELIF/ELSE - WHILE

# Datos iniciales

vel_l = 5  # velocidad máxima de la liebre
vel_t = 3  # velocidad máxima de la tortuga
l_trampas = [4, 7, 9, 13]



pos_liebre = 0
pos_tortuga = 0


for turno in range(1, 11):  # Del 1 al 10, se pone 11 porque no se incluye el final!!!
  print(f"El turno es: {turno}")

  pos_anterior_libre = pos_liebre

  avance_liebre = random.randint(1, vel_l)

  pos_liebre += avance_liebre

  # Verificar si la liebre cayó en una trampa!!

  if pos_liebre in l_trampas:
    print(f"Liebre cayó en trampa en casilla {pos_liebre}, regresa a su posición anterior")
    pos_liebre = pos_anterior_libre

  # VAMOS CON LA TORTUGUITA

  avance_tortuga = random.randint(1, vel_t)

  pos_tortuga += avance_tortuga

  #Ya avanzó la tortuga y la liebre, así que vamos a imprir los AVANCES

  guiones = "-"
  print(f"Liebre: {"-"*pos_liebre} (Posicion {pos_liebre})") # X ejemplo, Liebre pos = 5, se multiplica - 5 veces
  print(f"Tortuga: {"-"*pos_tortuga} (Posicion) {pos_tortuga}")


print(f"El ganador de la carrera es")

if pos_tortuga > pos_liebre:
  print("Tortuga!!")

elif pos_liebre > pos_tortuga:
  print("Liebre!!")

else: # Que ambas posiciones sean iguales
  print("Empate!!")









```
