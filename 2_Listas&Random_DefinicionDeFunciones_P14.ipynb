<h1>Funciones de <code>random</code> para listas</h1>

<h2>1) <code>random.choice(lista)</code></h2>
<p>Devuelve <b>un solo elemento aleatorio</b> de la lista (sin modificarla).</p>
<pre><code class="language-python">
import random
colores = ["rojo","verde","azul"]
print(random.choice(colores))
</code></pre>

<hr>

<h2>2) <code>random.sample(lista, k)</code></h2>
<p>Devuelve una <b>nueva lista</b> con <code>k</code> elementos <b>sin repetirse</b>.</p>
<pre><code class="language-python">
import random
ganadores = random.sample(colores, k=2)
print(ganadores)
</code></pre>

<hr>

<h2>4) <code>random.shuffle(lista)</code></h2>
<p>Mezcla los elementos <b>en la misma lista</b> (modifica la original).</p>
<pre><code class="language-python">
import random
nombres = ["Ana","Luis","Sofía","Marcos"]
random.shuffle(nombres)
print(nombres)
</code></pre>

<hr>

<p><b>Resumen rápido:</b></p>
<ul>
  <li><code>choice()</code> → un elemento</li>
  <li><code>choices()</code> → varios, con repetidos</li>
  <li><code>sample()</code> → varios, sin repetir</li>
  <li><code>shuffle()</code> → mezcla la lista original</li>
</ul>

<h2>Ejercicio 1.2 — Reparto aleatorio de equipos con <code>random</code> y listas</h2>

<p><b>Enunciado:</b></p>

<p>
Vas a crear un programa que reparta estudiantes en 3 equipos de forma completamente aleatoria.  
Cada estudiante pertenece a un solo equipo, y los equipos deben tener <b>el mismo número de integrantes</b> (o diferir máximo en 1 persona).  
No puedes usar bucles ni condicionales: solo <b>funciones de listas</b> y <b>funciones de <code>random</code></b>.
</p>

<ol>
  <li>Crea una lista llamada <code>estudiantes</code> con al menos 9 nombres.</li>
  <li>Desordena la lista usando <code>random.shuffle()</code>.</li>
  <li>Usa <code>len()</code> y slicing para dividir la lista en tres sublistas:
    <ul>
      <li><code>equipo1</code> → primeros 3 elementos</li>
      <li><code>equipo2</code> → siguientes 3</li>
      <li><code>equipo3</code> → últimos 3</li>
    </ul>
  </li>
  <li>De cada equipo, elige un <b>líder al azar</b> usando <code>random.choice()</code>.</li>
  <li>Selecciona 2 suplentes globales de entre todos los estudiantes restantes usando <code>random.sample()</code>.</li>
  <li>Muestra en pantalla:
    <ul>
      <li>Los tres equipos con sus integrantes.</li>
      <li>El líder de cada equipo.</li>
      <li>Los dos suplentes globales.</li>
    </ul>
  </li>
</ol>


```python
import random


estudiantes = ["Ana", "Luis", "Sofía", "Marcos", "Carla", "Pedro", "Lucía", "Andrés", "Daniel"]

copia = estudiantes.copy() # - Copia de SEGURIDAD

# Desordenar la LISTA
random.shuffle(copia)

total = len(copia)

tam_equipo = total // 3   #Division entera para tener la cantidad de integrantes

inici1 = tam_equipo
inicio2 = tam_equipo * 2

equipo1 = copia[:inici1]  # -- > lista[INICIO : FINAL : PASO] -- > Lista[0 :  INICIO1 : 1 ]
equipo2 = copia[inici1:inicio2]
equipo3 = copia[inicio2:]


# Agarar una sola persona al azar --> CHOICE

lider1 = random.choice(equipo1)
lider2 = random.choice(equipo2)
lider3 = random.choice(equipo3)

# Me aseguro que los lideres ya no puedan ser suplentes

copia.remove(lider1)
copia.remove(lider2)
copia.remove(lider3)

# Sumplentes --> CHOICE y agarra 1 x 1 || --> sample y agarra 2 en una linea

suplentes = random.sample(copia, k=2) # --> random.sample( LISTA , k = CANTIDAD DE ELEMENTOS) Y no se repiten

print(f" Equipo 1 : {equipo1}, el LIDER es {lider1}")
print(f" Equipo 2 : {equipo2}, el LIDER es {lider2}")
print(f" Equipo 3 : {equipo3}, el LIDER es {lider3}")

print(f" Los suplentes escogidos son {suplentes}")
```

<h2> Ejercicio 2.2
 </h2>

<p><b>Enunciado:</b></p>

<p>
Define una función llamada <code>analisis_aleatorio()</code> que trabaje con una lista de números y devuelva información,
usando funciones de <code>listas</code> y <code>random</code>.
</p>

<ol>
  <li>La función debe recibir dos parámetros:
    <ul>
      <li><code>valores</code>: una lista de números (por ejemplo, notas de estudiantes).</li>
      <li><code>n</code>: la cantidad de elementos que se seleccionarán al azar.</li>
    </ul>
  </li>
  <li>Dentro de la función:
    <ul>
      <li>Crea una copia de la lista original con <code>copy()</code>.</li>
      <li>Desordénala usando <code>random.shuffle()</code>.</li>
      <li>Selecciona <code>n</code> elementos aleatorios SIN repetirse usando <code>random.sample()</code>.</li>
      <li>Calcula el promedio de los seleccionados usando <code>sum()</code> y <code>len()</code>.</li>
      <li>Devuelve tres resultados con <code>return</code>:
        <ol type="a">
          <li>La lista original (sin modificar).</li>
          <li>La lista de valores seleccionados.</li>
          <li>El promedio de los seleccionados.</li>
        </ol>
      </li>
    </ul>
  </li>
  <li>Fuera de la función:
    <ul>
      <li>Crea una lista de 10 números entre 1 y 10 (pueden ser enteros o decimales).</li>
      <li>Llama a la función con <code>n = 4</code>.</li>
      <li>Guarda el resultado en tres variables y muestra los valores en pantalla.</li>
    </ul>
  </li>
</ol>


```python
import random

def analisis_aleatorio(valores, n):
  valores_copia = valores.copy()
  random.shuffle(valores_copia)

  seleccionados = random.sample(valores_copia, k = n)
  # --> De la lista VALORES COPIA, AGARRES ALEATORIAMENTE n elementos --> random.sample(valores_copia, k = n)

  promedio = sum(seleccionados)/len(seleccionados)

  return valores, seleccionados, promedio

# NO IDENTACION -- > ESTAS AFUERA DE LA FUNCION

# NO OLVIDAR -- Identacion (Espacio) -- DOS PUNTOS ":" -- "def"

chocapic = [2, 4, 6, 8, 10, 3, 5, 7, 9, 1]

lista_num, suertudos, promedio = analisis_aleatorio(chocapic, 4)

print(f"La lista original es {lista_num}")

print(f"Los numeros elegidos al azar son: {suertudos}")

print(f"El promedio de la lista es {promedio}")



```

<h2>Ejercicio 2.3</h2>

<p><b>Enunciado:</b></p>

<p>
La estación meteorológica “ClimaTEC” registra temperaturas diarias (en °C) durante una semana.  
Tu tarea es crear un conjunto de funciones para analizar los datos de manera automática usando listas, random y funciones.
</p>

<ol>
  <li>Define una función llamada <code>generar_temperaturas()</code> que reciba:
    <ul>
      <li><code>cantidad</code>: número de días a registrar.</li>
      <li><code>rango</code>: tupla con el rango de temperaturas posibles (por defecto <code>(15, 35)</code>).</li>
    </ul>
    La función debe:
    <ul>
      <li>Usar <code>random.choices()</code> para generar una lista de temperaturas enteras dentro del rango.</li>
      <li>Retornar la lista generada.</li>
    </ul>
  </li>

  <li>Define una segunda función llamada <code>analisis_temperaturas()</code> que reciba una lista de temperaturas y devuelva:
    <ul>
      <li>El valor máximo.</li>
      <li>El valor mínimo.</li>
      <li>El promedio de las temperaturas.</li>
    </ul>
    Usa <code>max()</code>, <code>min()</code> y <code>sum()/len()</code>.  
    Retorna los tres valores usando <code>return</code>.</li>

  <li>Crea una función booleana llamada <code>esta_en_rango(valor, rango)</code>  
      que devuelva <code>True</code> si la temperatura está dentro del rango indicado, y <code>False</code> en caso contrario.  
      Usa únicamente el operador <code>in</code> con <code>range()</code>.</li>

  <li>Fuera de las funciones:
    <ul>
      <li>Genera una lista de temperaturas para 7 días usando <code>generar_temperaturas()</code>.</li>
      <li>Usa <code>analisis_temperaturas()</code> para obtener el máximo, mínimo y promedio.</li>
      <li>Verifica con <code>esta_en_rango()</code> si una temperatura específica (por ejemplo 28) está dentro del rango generado.</li>
    </ul>
  </li>
</ol>

<p><i>Restricciones:</i>  
No usar bucles (<code>for</code>, <code>while</code>), condicionales (<code>if</code>), ni estructuras como <code>set()</code> o list comprehensions.</p>


```python
import random

# --- Función 1: Generar lista de temperaturas aleatorias ---
def generar_temperaturas(cantidad, rango=(15, 35)):
    # random.choices genera lista de enteros dentro del rango dado
    temperaturas = random.choices(range(rango[0], rango[1] + 1), k=cantidad)
    return temperaturas

# --- Función 2: Analizar temperaturas ---
def analisis_temperaturas(lista_temperaturas):
    temp_max = max(lista_temperaturas)
    temp_min = min(lista_temperaturas)
    promedio = sum(lista_temperaturas) / len(lista_temperaturas)
    return temp_max, temp_min, promedio

# --- Función 3: Booleana para verificar si un valor está dentro del rango ---
def esta_en_rango(valor, rango):
    return valor in range(rango[0], rango[1] + 1)

# --- Ejemplo de uso ---
dias = 7
temperaturas = generar_temperaturas(dias)

maxima, minima, promedio = analisis_temperaturas(temperaturas)
verificacion = esta_en_rango(28, (15, 35))

print("Temperaturas registradas:", temperaturas)
print("Máxima:", maxima, "°C")
print("Mínima:", minima, "°C")
print("Promedio:", round(promedio, 2), "°C")
print("¿La temperatura 28°C está dentro del rango?:", verificacion)
```

  <li>Define una función llamada <code>analizar_notas()</code> que reciba una lista de calificaciones (entre 0 y 10) y devuelva tres valores:

      <li>El promedio de las notas.</li>
      <li>La nota más alta.</li>

    Usa las funciones <code>sum()</code>, <code>len()</code>, <code>max()</code> y <code>min()</code>.</li>

  <li>Define otra función booleana llamada <code>nota_aprobatoria(valor)</code> que devuelva <code>True</code> si la nota está en el rango de aprobación (7 a 10) y <code>False</code> en caso contrario.  
      No uses <code>if</code>; trabaja con la expresión <code>valor in range(7, 11)</code>.</li>

  <li>Fuera de las funciones:
    <ul>
      <li>Crea una lista llamada <code>notas</code> con al menos 6 calificaciones.</li>

```python
def analizar_notas(notas):
  promedio = sum(notas)/len(notas)
  nota_max = max(notas)
  nota_min = min(notas)

  return promedio, nota_max, nota_min

def nota_aprobatorio(valor):
  return valor in range(7 , 11)


calificaciones = [8.5, 6.0, 9.0, 7.5, 5.8, 10.0]

promedio, nota_max, nota_min = analizar_notas(calificaciones)

aprobacion = nota_aprobatorio(promedio)

print(f"El promedio de las notas es: {int(promedio)}")
print(f"La nota mas alta es: {nota_max}")
print(f"La nota mas baja es: {nota_min}")
```
