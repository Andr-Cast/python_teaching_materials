<h2>1. Manejo de archivos CSV</h2>

<h3>1.1 Lectura de archivos con <code>pd.read_csv()</code></h3>

<p>
El método <code>pd.read_csv()</code> permite leer archivos en formato CSV
(Comma Separated Values) y cargarlos en un <strong>DataFrame</strong>,
estructura fundamental de Pandas para el análisis de datos.
</p>

<p><strong>Sintaxis básica:</strong></p>

<pre>
import pandas as pd
df = pd.read_csv("archivo.csv")
</pre>

<p><strong>Argumentos útiles:</strong></p>

<ul>
  <li>
    <strong><code>sep</code></strong>: indica el separador de columnas del archivo.
    <ul>
      <li><code>","</code> coma (por defecto)</li>
      <li><code>";"</code> punto y coma</li>
      <li><code>"\t"</code> tabulador</li>
    </ul>
    <pre>
df = pd.read_csv("datos.csv", sep=";")
    </pre>
  </li>

  <li>
    <strong><code>header</code></strong>: define la fila que contiene los nombres de las columnas.
    <ul>
      <li><code>header=0</code>: la primera fila es el encabezado</li>
      <li><code>header=None</code>: el archivo no tiene encabezados</li>
    </ul>
    <pre>
df = pd.read_csv("datos.csv", header=None)
    </pre>
  </li>

  <li>
    <strong><code>index_col</code></strong>: especifica qué columna se utilizará como índice del DataFrame.
    <pre>
df = pd.read_csv("datos.csv", index_col=0)
    </pre>
  </li>
</ul>

<h3>1.2 Escritura de archivos con <code>to_csv()</code></h3>

<p>
El método <code>to_csv()</code> permite guardar un DataFrame en un archivo CSV.
</p>

<pre>
df.to_csv("salida.csv")
</pre>

<p><strong>Argumentos frecuentes:</strong></p>

<ul>
  <li><code>index=False</code>: evita guardar el índice como una columna adicional.</li>
  <li><code>sep</code>: define el separador del archivo de salida.</li>
</ul>

<pre>
df.to_csv("salida.csv", index=False, sep=";")
</pre>

<h2>2. Operaciones aritméticas y lógicas</h2>

<h3>2.1 Operaciones aritméticas entre Series</h3>

<p>
Las <strong>Series</strong> de Pandas permiten realizar operaciones matemáticas
elemento a elemento de forma vectorizada.
</p>

<pre>
import pandas as pd

a = pd.Series([10, 20, 30])
b = pd.Series([1, 2, 3])

suma = a + b
resta = a - b
producto = a * b
division = a / b
</pre>

<p>
Las operaciones se realizan alineando los valores según su índice.
</p>

<h3>2.2 Operaciones booleanas con operadores relacionales</h3>

<p>
Los operadores relacionales generan Series booleanas con valores
<code>True</code> y <code>False</code>.
</p>

<pre>
edades = pd.Series([12, 18, 25, 10])
mayores_edad = edades >= 18
</pre>

<p><strong>Operadores más utilizados:</strong></p>

<ul>
  <li><code>&gt;</code> mayor que</li>
  <li><code>&lt;</code> menor que</li>
  <li><code>&gt;=</code> mayor o igual</li>
  <li><code>&lt;=</code> menor o igual</li>
  <li><code>==</code> igual</li>
  <li><code>!=</code> diferente</li>
</ul>

<h3>2.3 Filtros booleanos usando <code>&amp;</code> y <code>|</code></h3>

<p>
Para combinar condiciones en Pandas se utilizan los operadores:
</p>

<ul>
  <li><code>&amp;</code> para AND</li>
  <li><code>|</code> para OR</li>
</ul>

<p>
No se deben usar <code>and</code> ni <code>or</code>.  
Cada condición debe ir entre paréntesis.
</p>

<pre>
df_filtrado = df[(df["edad"] >= 18) & (df["nota"] >= 14)]
</pre>

<pre>
df_filtrado = df[(df["ciudad"] == "Quito") | (df["ciudad"] == "Guayaquil")]
</pre>

<h3>2.4 Uso del método <code>.mask()</code></h3>

<p>
El método <code>.mask()</code> reemplaza los valores de una Series o DataFrame
cuando se cumple una condición.
</p>

<pre>
notas = pd.Series([10, 15, 18, 12])
resultado = notas.mask(notas < 14, "Reprobado")
</pre>

<p>
Si la condición es verdadera, el valor se reemplaza;  
si es falsa, el valor original se mantiene.
</p>

<h3>2.5 Métodos <code>.idxmax()</code> y <code>.idxmin()</code></h3>

<p>
Estos métodos devuelven el <strong>índice</strong> del valor máximo o mínimo
de una Series.
</p>

<pre>
ventas = pd.Series([100, 250, 180], index=["Enero", "Febrero", "Marzo"])

mes_mayor_venta = ventas.idxmax()
mes_menor_venta = ventas.idxmin()
</pre>

<p>
Son útiles para identificar rápidamente valores extremos sin necesidad de ordenar los datos.
</p>

<h2>Ejercicio 2: Operaciones aritméticas e identificación de valores extremos</h2>

<p>
Se tienen las ventas mensuales de dos productos almacenadas en Series de Pandas.
</p>

<p>Realice las siguientes actividades:</p>

<ol>
  <li>Cree dos Series llamadas <code>ventas_A</code> y <code>ventas_B</code>.</li>
  <li>Calcule la Series <code>ventas_totales</code> sumando ambas Series.</li>
  <li>Identifique el mes con mayores ventas totales.</li>
  <li>Identifique el mes con menores ventas totales.</li>
</ol>


```python
# Primero se crean las series de Pandas.
# Si vemos SERIES o DATAFRAME --> IMPORTAR pandas

import pandas as pd

serie_ventas_A = pd.Series([100,150,100,180], index = ["Enero", "Febrero", "Marzo", "Abril"])
serie_ventas_B = pd.Series([80,90,110,50],index = ["Enero", "Febrero", "Marzo", "Abril"])

# Las SERIES deben de tener la MISMA cantidad de DATOS!!

print(f"Las ventas A son:")
print(serie_ventas_A)
print(f"\nLas ventas B son:")
print(serie_ventas_B)

serie_ventas_totales = serie_ventas_A + serie_ventas_B

print(f"\nLas ventas totales son:")
print(serie_ventas_totales)

# RECORDAR 1. Importar pandas - 2. Usar Mayusculas y Minusculas - 3. Longitud del indice == cantidad de datos

mes_mas_alto = serie_ventas_totales.idxmax()

print(f"\nEl mes con mas ventas es: {mes_mas_alto}")

mes_mas_bajo = serie_ventas_totales.idxmin()

print(f"El mes con menos ventas es: {mes_mas_bajo}")

```

<h2>Ejercicio 3: Filtros booleanos y uso de mask()</h2>

<p>
Se cuenta con un conjunto de datos de estudiantes con sus edades y calificaciones.
</p>

<p>Realice las siguientes actividades:</p>

<ol>
  <li>Cree un DataFrame con las columnas <code>nombre</code>, <code>edad</code> y <code>nota</code>.</li>
  <li>Filtre a los estudiantes con edad mayor o igual a 18 y nota mayor o igual a 14.</li>
  <li>Cree una nueva columna <code>estado</code> usando <code>.mask()</code>.</li>
  <li>Muestre el DataFrame final.</li>
</ol>

```python
import pandas as pd

datos = {
    "Nombre": ["Ana", "Luis", "María", "Carlos", "Sofía"],
    "Edad": [17, 18, 20, 16, 22],
    "Nota": [15, 12, 18, 10 ,14]
}


df = pd.DataFrame(datos)

print(df)

df_filtrado = df[(df["Edad"] >= 18) & (df["Nota"] >= 14)]

print(f"\nEstudiantes con edad mayor o igual a 18 y nota mayor o igual a 14:")
print(df_filtrado)

df["Estado"] = "Aprobado"

print(f"\nLos estudiantes aprobados son")

df["Estado"] = df["Estado"].mask(df["Nota"] < 14, "Reprobado" )

print(df)
```

<h2>Ejercicio 4: Manejo de CSV, filtros y idxmax()</h2>

<p>
Se dispone de un archivo <strong>ventas.csv</strong> con las columnas
<em>mes</em>, <em>vendedor</em> y <em>ventas</em>.
</p>

<p>Realice las siguientes actividades:</p>

<ol>
  <li>Lea el archivo CSV usando <code>pd.read_csv()</code>.</li>
  <li>Filtre las filas con ventas mayores o iguales a 500.</li>
  <li>Determine el mes con la mayor venta usando <code>.idxmax()</code>.</li>
  <li>Guarde el DataFrame filtrado en un nuevo archivo CSV sin el índice.</li>
</ol>

```python
import pandas as pd

df = pd.read_csv("ventas.csv")
print(df)

df_filtrado = df[df["Ventas"] >= 500]

# El mes con mayor venta --- > INDEX

mes_mayor_venta = df_filtrado["Ventas"].idxmax()

print(f"\nEl mes con mayor venta es: {mes_mayor_venta}")

# Para guardar o CREAR archivos de PANDAS a CSV se usa el "to_csv"
# Necesitamos un DATA FRAMSE preexistente

df_filtrado.to_csv("ventas_filtradas.csv", index = False)
```
