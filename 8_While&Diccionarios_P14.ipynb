<h1>Ejercicio sencillo: Diccionario de precios de frutas</h1>

<p>
Escribe un programa en Python que:
</p>

<ol>
  <li>Defina un diccionario llamado <code>precios</code> con al menos 3 frutas y sus precios por kilo. Por ejemplo:
    <pre>
precios = {
    "manzana": 1500,
    "banana": 1200,
    "naranja": 1800
}
    </pre>
  </li>
  <li>Le pida al usuario que ingrese el nombre de una fruta.</li>
  <li>Busque esa fruta en el diccionario:
    <ul>
      <li>Si la fruta existe, muestre el precio por kilo.</li>
      <li>Si la fruta no existe, muestre un mensaje indicando que no está disponible.</li>
    </ul>
  </li>
</ol>

<p>
Considere que el usuario puede escribir la fruta con mayúsculas o minúsculas. El programa debe funcionar correctamente en ambos casos.
</p>


```python
# Crear un diccionario de Frutas-Precios/Kg


# Clave - Fruta
# Valor - Precio/kg

precios = {
    "manzana" : 1500,
    "banana"  : 1000,
    "naranja" : 1200
}

# Nombre del diccionario
# Llaves {}
# Claves : Valores,




fruta = input(f"Ingrese el nombre de una fruta: ")      # Cuando hacemos input, hay que NORMALIZAR

fruta_normalizada = fruta.strip().lower()

if fruta_normalizada in precios:
  print(f"El precio por kg de {fruta} es de ${precios[fruta_normalizada]}")
else:
  print(f"La fruta no esta registrada")



```

<h1>Ejercicio: Registro de notas con diccionarios</h1>

<p>
Se desea un programa en Python que administre las notas de un curso. Cada estudiante se identifica por su nombre, y se desea almacenar una única nota final para cada uno.
</p>

<p>
El programa debe utilizar un diccionario llamado <code>notas</code>, donde:
</p>
<ul>
  <li>La clave es el nombre del estudiante (cadena de texto).</li>
  <li>El valor es la nota final (número).</li>
</ul>

<p>
El programa principal debe mostrar un menú de opciones que se repite hasta que el usuario elija salir. El menú debe permitir, al menos:
</p>

<ol>
  <li>Registrar o actualizar la nota de un estudiante, a partir de su nombre.</li>
  <li>Consultar la nota de un estudiante, ingresando su nombre.</li>
  <li>Mostrar la cantidad de estudiantes registrados y el promedio de sus notas.</li>
  <li>Salir del programa.</li>
</ol>

```python
def mostrar_menu():
  print(f"El menu de opciones es:")
  print(f"1. Registrar o actualizar la nota de un estudiante")
  print(f"2. Consultar la nota de un estudiante, ingresando su nombre.")
  print(f"3. Mostrar la cantidad de estudiantes registrados y el promedio de sus notas.")
  print(f"4. Salir del programa.")

# Diccionario con Calificaciones: "diccionario_notas"

def registrar_actualizar_nota(diccionario_notas):
  # Esta funcion permitirá AGREGAR o ACTUALIZAR una calificacion
  # Necesito una CLAVE y un VALOR
  # Clave - Nombre / Valor - Calificacion

  nombre = input(f"Ingrese el nombre del estudiante:").strip().lower()
  # Ya tengo mi Clave, ahora agrego el VALOR

  nota = float(input(f"Ingrese la nota final del estudiante"))

  # Ya tengo la CLAVE y el VALOR!!
  diccionario_notas[nombre] = nota

  # DOS CASOS
  # 1 - Ya EXISTA el estudiante. --> Se ACTUALIZA por el nuevo Valor
  # 2 - NO existe el estudiante. --> Se va a CREAR un nuevo estudiante



def consultar_nota(diccionario_notas):
  # Imprimir en pantalla LA CALIFICACION de un estudiante

  nombre = input(f"¿Que estudiante desea consultar?").lower().strip()

  calificacion = diccionario_notas[nombre]

  print(f"La calificacion  consultada es: {calificacion}")



def mostrar_cantidad_promedio(diccionario_notas):

  # 1. Contar la CANTIDAD de estudiantes registrados !!
  # 2. PROMEDIAR la calificacion de los estudiantes

  cantidad = len(diccionario_notas) # Cantidad de estudiantes!!

  # SUMAR todas las calificaciones y DIVIDIRLAS para la cantidad de estudiantes
  # .values()-- > Una tupla de solo VALORES (Calificaciones)

  tupla_calificaciones = diccionario_notas.values() # Tupla con las calificaciones de estudaintes
  promedio_calificaciones = sum(tupla_calificaciones)/cantidad  # Obtengo el promedio = suma/longitud

  print(f"La cantidad de estudiantes registrados es: {cantidad}")
  print(f"El promedio de calificaciones es: {promedio_calificaciones}")



# El usuario debe de poder ESCOGER entre las 4 opciones
# BUCLE donde el usuario pueda ver las opciones!!!
# OPCION - DICCIONARIO

notas = {
    "eduardo" : 10,
    "pedro"   : 20,
    "juan"    : 1,
}

opcion = "0" # VALOR INICIAL

# WHILE: VALOR INICIAL -> VALIDACION DE CONDICION -> ACTUALIZAMOS LA CONDICION

while opcion != "4":

  mostrar_menu()

  opcion = input(f"¿Qué desea hacer hoy? (1,2,3,4): ")

  if opcion == "1":
    registrar_actualizar_nota(notas)

  elif opcion == "2":
    consultar_nota(notas)

  elif opcion == "3":
    mostrar_cantidad_promedio(notas)

  else:
    opcion == "4"
    print(f"Se acaba el programa!!")
```

<h1>Ejercicio: Catálogo de películas con diccionarios</h1>

<p>
Se desea un programa en Python que administre un catálogo de películas. Cada película se identifica por su título y debe almacenar varios datos asociados.
</p>

<p>
El programa debe utilizar un diccionario llamado <code>peliculas</code>, donde:
</p>
<ul>
  <li>La clave es el título de la película (cadena de texto).</li>
  <li>El valor es otra estructura con información de la película (por ejemplo, un diccionario que incluya director y año, o una lista con esos datos).</li>
</ul>

<p>
El programa principal debe mostrar un menú de opciones que se repite hasta que el usuario elija salir. El menú debe permitir, al menos:
</p>

<ol>
  <li>Registrar una nueva película, ingresando su título y al menos dos datos asociados (por ejemplo, director y año).</li>
  <li>Buscar una película por su título y mostrar toda la información almacenada de ella.</li>
  <li>Mostrar todas las películas registradas con sus datos.</li>
  <li>Salir del programa.</li>
</ol>

<p>
La solución debe utilizar:
</p>
<ul>
  <li>Un diccionario donde cada valor contenga más de un dato sobre la película.</li>
  <li>Un ciclo <code>while</code> para el manejo del menú.</li>
  <li>Condicionales para decidir la acción según la opción ingresada.</li>
  <li>Al menos dos funciones definidas por el o la estudiante.</li>
  <li>Manejo de cadenas de texto al trabajar con los títulos de las películas.</li>
</ul>


```python
def mostrar_menu_peliculas():
    """Muestra el menú principal para el catálogo de películas."""
    print("\nMenú catálogo de películas")
    print("1. Registrar o actualizar película")
    print("2. Buscar película por título")
    print("3. Mostrar todas las películas")
    print("4. Salir")


def normalizar_titulo(titulo):
    return titulo.strip().lower()


def registrar_pelicula(peliculas):

    titulo = input("Ingresa el título de la película: ")
    clave = normalizar_titulo(titulo)

    director = input("Ingresa el nombre del director: ")
    anio = input("Ingresa el año de estreno: ")
    genero = input("Ingresa el género de la película: ")

    # Se almacena un diccionario como valor, con varios datos
    peliculas[clave] = {
        "titulo": titulo.strip(),
        "director": director.strip(),
        "anio": anio.strip(),
        "genero": genero.strip()
    }

    print("Película registrada o actualizada correctamente.")


def buscar_pelicula(peliculas):
    """
    Busca una película en el diccionario a partir del título ingresado.
    Muestra la información si la encuentra.
    """
    if len(peliculas) == 0:
        print("No hay películas registradas.")
        return

    titulo = input("Ingresa el título de la película que deseas buscar: ")
    clave = normalizar_titulo(titulo)

    if clave in peliculas:
        datos = peliculas[clave]
        print("\nInformación de la película:")
        print("Título :", datos["titulo"])
        print("Director:", datos["director"])
        print("Año    :", datos["anio"])
        print("Género :", datos["genero"])
    else:
        print("La película no se encuentra en el catálogo.")


def mostrar_peliculas(peliculas):
    """
    Muestra todas las películas almacenadas en el diccionario.
    Recorre solo los valores, que son los diccionarios con datos.
    """
    if len(peliculas) == 0:
        print("No hay películas registradas.")
        return

    print("\nCatálogo de películas:")
    print("-----------------------")
    for datos in peliculas.values():
        print("Título :", datos["titulo"])
        print("Director:", datos["director"])
        print("Año    :", datos["anio"])
        print("Género :", datos["genero"])
        print("-----------------------")


# Programa principal para el ejercicio de películas
peliculas = {}
opcion_peliculas = "0"  # valor inicial distinto de "4" para entrar al while

while opcion_peliculas != "4":
    mostrar_menu_peliculas()
    opcion_peliculas = input("Elige una opción: ").strip()

    if opcion_peliculas == "1":
        registrar_pelicula(peliculas)
    elif opcion_peliculas == "2":
        buscar_pelicula(peliculas)
    elif opcion_peliculas == "3":
        mostrar_peliculas(peliculas)
    elif opcion_peliculas == "4":
        print("Fin del programa de catálogo de películas.")
    else:
        print("Opción no válida. Intenta nuevamente.")
```

<h1>Ejercicio: Registro de pacientes en una consulta médica</h1>

<p>
Se desea implementar en Python un programa que administre la información de los pacientes de una consulta médica.
</p>

<p>
El programa debe utilizar un diccionario llamado <code>pacientes</code>, donde:
</p>
<ul>
  <li>La clave es un identificador de paciente (por ejemplo, RUT o número de ficha, como cadena de texto).</li>
  <li>El valor es una estructura que almacene al menos: nombre del paciente, edad y motivo de la consulta (por ejemplo, otro diccionario o una lista).</li>
</ul>

<p>
El programa principal debe mostrar un menú de opciones que se repite hasta que el usuario elija salir. El menú debe permitir, al menos:
</p>

<ol>
  <li>Registrar un paciente nuevo o actualizar los datos de un paciente existente, ingresando su identificador.</li>
  <li>Buscar un paciente por su identificador y mostrar todos sus datos almacenados.</li>
  <li>Mostrar la cantidad total de pacientes registrados y listar sus identificadores y nombres.</li>
  <li>Eliminar un paciente del registro a partir de su identificador.</li>
  <li>Salir del programa.</li>
</ol>

<p>
La solución debe utilizar un diccionario con valores que contengan más de un dato por paciente, un ciclo <code>while</code> para el manejo del menú, condicionales, al menos dos funciones definidas por el o la estudiante y manejo de cadenas de texto al trabajar con identificadores y nombres.
</p>


```python
def mostrar_menu_pacientes():
    """Muestra el menú principal para el registro de pacientes."""
    print("\nMenú registro de pacientes")
    print("1. Registrar o actualizar paciente")
    print("2. Buscar paciente por identificador")
    print("3. Mostrar cantidad de pacientes y listado")
    print("4. Eliminar paciente")
    print("5. Salir")


def normalizar_identificador(identificador):
    """
    Normaliza el identificador del paciente:
    - elimina espacios al inicio y al final
    - pasa a mayúsculas (por si hay letras en el RUT u otro código)
    """
    return identificador.strip().upper()


def registrar_paciente(pacientes):
    """
    Registra un nuevo paciente o actualiza los datos de uno existente.
    La clave del diccionario es el identificador normalizado.
    El valor es un diccionario con nombre, edad y motivo de consulta.
    """
    identificador = input("Ingresa el identificador del paciente: ")
    clave = normalizar_identificador(identificador)

    nombre = input("Ingresa el nombre del paciente: ")
    edad = int(input("Ingresa la edad del paciente: "))
    motivo = input("Ingresa el motivo de consulta: ")

    pacientes[clave] = {
        "nombre": nombre.strip(),
        "edad": edad,
        "motivo": motivo.strip()
    }

    print("Paciente registrado o actualizado correctamente.")


def buscar_paciente(pacientes):
    """
    Busca un paciente en el diccionario a partir de su identificador.
    Muestra todos los datos si lo encuentra.
    """
    if len(pacientes) == 0:
        print("No hay pacientes registrados.")
        return

    identificador = input("Ingresa el identificador del paciente: ")
    clave = normalizar_identificador(identificador)

    if clave in pacientes:
        datos = pacientes[clave]
        print("\nDatos del paciente:")
        print("Identificador:", clave)
        print("Nombre       :", datos["nombre"])
        print("Edad         :", datos["edad"])
        print("Motivo       :", datos["motivo"])
    else:
        print("El paciente no se encuentra registrado.")


def mostrar_resumen_pacientes(pacientes):
    """
    Muestra la cantidad total de pacientes y
    lista sus identificadores y nombres.
    """
    if len(pacientes) == 0:
        print("No hay pacientes registrados.")
        return

    print("\nResumen de pacientes registrados")
    print("Cantidad total:", len(pacientes))
    print("--------------------------------")

    for identificador, datos in pacientes.items():
        print("Identificador:", identificador, "- Nombre:", datos["nombre"])


def eliminar_paciente(pacientes):
    """
    Elimina un paciente del diccionario a partir de su identificador.
    Si el identificador no existe, muestra un mensaje.
    """
    if len(pacientes) == 0:
        print("No hay pacientes registrados.")
        return

    identificador = input("Ingresa el identificador del paciente a eliminar: ")
    clave = normalizar_identificador(identificador)

    if clave in pacientes:
        del pacientes[clave]
        print("Paciente eliminado correctamente.")
    else:
        print("No existe un paciente con ese identificador.")


# Programa principal para el ejercicio de pacientes
pacientes = {}
opcion_pacientes = "0"  # valor inicial distinto de "5" para entrar al while

while opcion_pacientes != "5":
    mostrar_menu_pacientes()
    opcion_pacientes = input("Elige una opción: ").strip()

    if opcion_pacientes == "1":
        registrar_paciente(pacientes)
    elif opcion_pacientes == "2":
        buscar_paciente(pacientes)
    elif opcion_pacientes == "3":
        mostrar_resumen_pacientes(pacientes)
    elif opcion_pacientes == "4":
        eliminar_paciente(pacientes)
    elif opcion_pacientes == "5":
        print("Fin del programa de registro de pacientes.")
    else:
        print("Opción no válida. Intenta nuevamente.")
```
