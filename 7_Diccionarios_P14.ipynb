<h1>Diccionarios en Python</h1>

<h2>1. ¿Qué es un diccionario?</h2>
<p>
Un <strong>diccionario</strong> en Python es una estructura de datos que almacena información en forma de
<strong>pares clave–valor</strong>. A diferencia de las listas, donde accedemos por índice numérico, en un
diccionario accedemos a los datos usando una <strong>clave</strong>.
</p>

<p>Ejemplos de situaciones donde usar diccionarios:</p>
<ul>
  <li>Estudiantes: RUT → datos del estudiante.</li>
  <li>Productos: código → nombre, precio, stock.</li>
  <li>Traducciones: palabra en español → palabra en inglés.</li>
</ul>

<hr>

<h2>2. Sintaxis básica</h2>
<p>La sintaxis general de un diccionario es:</p>

<pre>
nombre_diccionario = {
    "clave1": valor1,
    "clave2": valor2,
    "clave3": valor3
}
</pre>

<p>
Las <strong>claves</strong> suelen ser <code>str</code> (cadenas), pero también pueden ser números u otros tipos
inmutables. Los <strong>valores</strong> pueden ser de cualquier tipo: números, cadenas, listas, otros diccionarios, etc.
</p>

<hr>

<h2>3. Crear diccionarios</h2>

<h3>3.1. Diccionario vacío</h3>
<pre>
datos = {}
</pre>

<h3>3.2. Diccionario con datos iniciales</h3>
<pre>
estudiante = {
    "nombre": "Ana",
    "edad": 20,
    "carrera": "Informática"
}
</pre>

<hr>

<h2>4. Acceder a valores</h2>
<p>Se usa la clave entre corchetes:</p>

<pre>
print(estudiante["nombre"])   # "Ana"
print(estudiante["edad"])     # 20
</pre>

<p><strong>Importante:</strong> si la clave no existe, <code>diccionario["clave"]</code> produce error.</p>

<h3>4.1. Uso de <code>get</code></h3>
<p><code>get</code> permite acceder a una clave y, si no existe, retornar un valor por defecto.</p>

<pre>
print(estudiante.get("nombre"))           # "Ana"
print(estudiante.get("nota", "No existe"))  # "No existe"
</pre>

<hr>

<h2>5. Agregar y modificar elementos</h2>

<h3>5.1. Agregar una nueva clave</h3>
<pre>
estudiante["nota_final"] = 5.5
</pre>

<h3>5.2. Modificar un valor existente</h3>
<pre>
estudiante["edad"] = 21
</pre>

<hr>

<h2>6. Eliminar elementos</h2>

<h3>6.1. Usando <code>del</code></h3>
<pre>
del estudiante["nota_final"]
</pre>

<h3>6.2. Usando <code>pop</code></h3>
<pre>
edad = estudiante.pop("edad")   # elimina la clave "edad" y retorna su valor
</pre>

<hr>

<h2>7. Recorrer diccionarios</h2>

<h3>7.1. Recorrer solo claves</h3>
<pre>
for clave in estudiante:
    print(clave)
</pre>

<h3>7.2. Recorrer claves explícitamente</h3>
<pre>
for clave in estudiante.keys():
    print(clave)
</pre>

<h3>7.3. Recorrer solo valores</h3>
<pre>
for valor in estudiante.values():
    print(valor)
</pre>

<h3>7.4. Recorrer clave y valor a la vez</h3>
<pre>
for clave, valor in estudiante.items():
    print(clave, "→", valor)
</pre>

<hr>

<h2>8. Operadores útiles</h2>

<h3>8.1. Verificar si una clave existe</h3>
<pre>
if "nombre" in estudiante:
    print("El estudiante tiene nombre registrado")
</pre>

<h3>8.2. Largo del diccionario</h3>
<pre>
print(len(estudiante))   # cantidad de pares clave–valor
</pre>

<hr>

<h2>9. Diccionarios anidados</h2>
<p>
Un valor puede ser otro diccionario. Esto es útil para representar datos más complejos, por ejemplo, varios estudiantes:
</p>

<pre>
estudiantes = {
    "12345678-9": {
        "nombre": "Ana",
        "edad": 20
    },
    "98765432-1": {
        "nombre": "Luis",
        "edad": 22
    }
}
</pre>

<p>Acceso a datos anidados:</p>

<pre>
print(estudiantes["12345678-9"]["nombre"])   # "Ana"
</pre>

<hr>

<h2>10. Diferencias principales entre listas y diccionarios</h2>
<ul>
  <li><strong>Lista:</strong> se accede por índice numérico (<code>lista[0]</code>).</li>
  <li><strong>Diccionario:</strong> se accede por clave (<code>diccionario["clave"]</code>).</li>
  <li>Las listas son útiles para colecciones ordenadas de elementos similares.</li>
  <li>Los diccionarios son útiles cuando cada elemento tiene una “etiqueta” o identificador.</li>
</ul>

<hr>

<h2>11. Resumen</h2>
<ul>
  <li>Un diccionario almacena pares <strong>clave–valor</strong>.</li>
  <li>Se define con <code>{}</code> y separando clave y valor con <code>:</code>.</li>
  <li>Permite agregar, modificar y eliminar elementos fácilmente.</li>
  <li>Es muy útil para representar objetos del mundo real con varios atributos.</li>
</ul>


<h1>Ejercicio: Registro básico de un estudiante con diccionarios</h1>

<ol>

```python
estudiante = {}

nombre = input(f"Ingrese el nombre del estudiante:") # El Usr agrega el nombre
edad = int(input(f"Ingrese la edad del Estudiante: "))
carrera = input(f"Ingrese la carrera del estudiante: ")

# INDEXAR las claves nuevas EN el diccionario y DARLES VALOR

estudiante["nombre"] = nombre # CREO la clave "nombre" y le ASIGNO el valor nombre
estudiante["edad"] = edad     # Lo mismo pero con edad
estudiante["carrera"] = carrera



nota_1 = float(input(f"Ingrese la primera calificacion: "))
nota_2 = float(input(f"Ingrese la Segunda calificacion: "))

lista_notas = [nota_1 , nota_2]

# CREAMOS la clave, y al mismo tiempo, ASIGNAMOS el valor

estudiante["notas"] = lista_notas

promedio = (nota_1+nota_2)/2

# AGREGAR EL PROMEDIO AL DICC

estudiante["promedio"] = promedio

print(f"\nResumen del estudiante")
print(f"----------------------------")

print(f"Nombre: {estudiante["nombre"]}")
print(f"Edad: {estudiante["edad"]}")
print(f"Carrera: {estudiante["carrera"]}")
print(f"Notas: {estudiante["notas"]}")
print(f"Promedio: {estudiante["promedio"]}")

# Kevin se dio cuenta pilas!! - No es necesario crealo vacio si se usará esto:
estudiantes = {
    "nombre" : nombre,
    "edad" : edad,
    "carrera" : carrera
}

# Igual recomiendo crearlo vacio!
```

<h1>Ejercicio 2: Diccionario de productos y precios</h1>

<h2>Objetivo</h2>
<p>
Practicar la creación, acceso, verificación y actualización de datos en un diccionario en Python.
</p>

<h2>Enunciado</h2>
<p>
Escribe un programa en Python que:
</p>
<ol>
  <li>Cree un diccionario llamado <code>precios</code> con al menos 3 productos iniciales. Por ejemplo:
    <pre>
precios = {
    "pan": 1500,
    "leche": 1200,
    "huevos": 2200
}
    </pre>
  </li>
  <li>Muestre en pantalla todos los productos disponibles.</li>
  <li>Pida al usuario que ingrese el nombre de un producto.</li>
  <li>Use el diccionario para:
    <ul>
      <li>Si el producto existe, mostrar su precio.</li>
      <li>Si el producto no existe, mostrar un mensaje indicando que no está registrado.</li>
    </ul>
  </li>
  <li>En caso de que el producto no exista:
    <ul>
      <li>Preguntar al usuario si desea agregarlo al diccionario.</li>
      <li>Si responde que sí, pedir el precio del nuevo producto (entero) y guardarlo en el diccionario.</li>
      <li>Al final, mostrar el diccionario completo de productos y precios.</li>
    </ul>
  </li>
</ol>


```python
precios = {
    "pan": 1500,
    "leche": 1200,
    "huevos": 2200
}

print(f"Los productos disponibles son:")
for producto in precios:  # ASÍ NOMAS, DEVUELVE LAS CLAVE!!
  print(f"- {producto}")


nombre_prod = input(f"Ingrese el nombre de un producto: ").strip().lower()

if nombre_prod in precios:
  print(f"El precio del producto seleccionado ({nombre_prod}) es: {precios[nombre_prod]} ")

else:
  #El producto NO esta en el diccionario de precios
  print(f"No existe el producto indicado")
  respuesta = input(f"Desea agregar el producto al diccionario? (si/no)").strip().lower()

  # texto = HOLA ESTAS - .strip() borra los espacios al inicio y al final, no los del centro!

  if respuesta == "si":
    # CREAR la clave y ASIGNAR el valor
    valor_nuevo = int(input(f"Ingrese el precio del producto a añadir"))

    precios[nombre_prod] = valor_nuevo
  else:
    print(f"No se agrega nadota al diccionario")


# RECORRER el diccionar e imprimir el PRODUCTO (clave) y el PRECIO (valor)

# Imprimir CLAVE Y VALOR

# .items()

print(f"\nEl Diccionario de precios y productos es:")
for producto, precio in precios.items():  # --> Primero CLAVE (producto) luego PRECIO (valor), usando ITEMS
  print(f"Producto: {producto} - Precio: {precio}")


```

<h1>Ejercicio 3: Diccionario de contactos telefónicos</h1>

<h2>Objetivo</h2>
<p>
Practicar la creación y actualización de diccionarios en Python, junto con la búsqueda de información a partir de una clave.
</p>

<h2>Enunciado</h2>
<p>
Escribe un programa en Python que simule una pequeña agenda de contactos:
</p>
<ol>
  <li>Cree un diccionario vacío llamado <code>agenda</code>.</li>
  <li>Use un ciclo <code>for</code> para pedir al usuario los datos de 3 contactos:
    <ul>
      <li>Nombre del contacto</li>
      <li>Teléfono del contacto</li>
    </ul>
  </li>
  <li>Guarde cada contacto en el diccionario usando:
    <ul>
      <li>Clave: el nombre (en minúsculas, sin espacios al inicio o final).</li>
      <li>Valor: el teléfono como cadena (string).</li>
    </ul>
  </li>
  <li>Al terminar de ingresar los 3 contactos, muestre todos los contactos registrados.</li>
  <li>Luego, pida al usuario el nombre de un contacto a buscar:
    <ul>
      <li>Si el contacto existe en la agenda, mostrar su número de teléfono.</li>
      <li>Si no existe, mostrar un mensaje indicando que no se encuentra registrado.</li>
    </ul>
  </li>
</ol>


```python
# Solución propuesta con comentarios para el profesor

# 1. Crear un diccionario vacío para la agenda
agenda = {}

# 2. Usar un ciclo for para pedir 3 contactos
#    En cada iteración se pide nombre y teléfono
for i in range(3):
    print("Contacto", i + 1)
    nombre = input("Ingresa el nombre del contacto: ")
    telefono = input("Ingresa el teléfono del contacto: ")

    # Normalizar el nombre:
    # strip() para eliminar espacios al inicio y final
    # lower() para guardar todo en minúsculas
    nombre_normalizado = nombre.strip().lower()

    # Guardar en el diccionario:
    # clave: nombre_normalizado
    # valor: teléfono (como string)
    agenda[nombre_normalizado] = telefono

    print("Contacto agregado.\n")

# 3. Mostrar todos los contactos registrados
print("Contactos registrados en la agenda:")
# Se recorre el diccionario con items() para obtener clave y valor
for nombre, telefono in agenda.items():
    # Se muestra el nombre tal como está guardado (minúsculas) y su teléfono
    print("Nombre:", nombre, "- Teléfono:", telefono)

# 4. Pedir al usuario el nombre de un contacto a buscar
nombre_buscar = input("\nIngresa el nombre del contacto a buscar: ")

# Normalizar de la misma forma que al guardar
nombre_buscar_normalizado = nombre_buscar.strip().lower()

# 5. Verificar si el contacto existe en la agenda
if nombre_buscar_normalizado in agenda:
    # Si existe, se accede directamente al teléfono usando la clave
    telefono_encontrado = agenda[nombre_buscar_normalizado]
    print("El teléfono de", nombre_buscar, "es:", telefono_encontrado)
else:
    # Si no existe, se muestra un mensaje de no encontrado
    print("El contacto", nombre_buscar, "no se encuentra registrado en la agenda.")
```

<h1>Ejercicio 4: Registro de productos con while y diccionarios</h1>

</p>Escribe un programa en Python que permita registrar productos y sus precios en un diccionario.
</p>
<ol>
  <li>Crea un diccionario vacío llamado <code>productos</code>.</li>
  <li>El programa debe pedir al usuario el nombre de un producto:
    <ul>
      <li>Si el usuario escribe <code>FIN</code> (sin importar mayúsculas/minúsculas), el programa deja de registrar productos.</li>
      <li>En caso contrario, se debe pedir el precio del producto (un entero positivo).</li>
      <li>Guarda el producto en el diccionario:
        <ul>
          <li>Clave: el nombre del producto en minúsculas y sin espacios al inicio ni al final.</li>
          <li>Valor: el precio del producto (entero).</li>
        </ul>
      </li>
      <li>Si el producto ya existe en el diccionario, muestra un mensaje indicando que ya está registrado y no lo vuelvas a agregar.</li>
    </ul>
  </li>
  <li>El registro de productos debe implementarse con un ciclo <code>while</code> que termine cuando el usuario escriba <code>FIN</code> (sin usar <code>break</code> ni <code>while True</code>).</li>
  <li>Al finalizar:
    <ul>
      <li>Si no se registró ningún producto, muestra un mensaje indicándolo.</li>
      <li>Si hay productos registrados:
        <ul>
          <li>Muestra todos los productos con sus precios.</li>
          <li>Determina y muestra cuál es el producto más caro y su precio.</li>
        </ul>
      </li>
    </ul>
  </li>
</ol>

<h2>Sugerencia</h2>
<pre>
productos = {}

nombre = input("Ingresa el nombre del producto (FIN para terminar): ")
nombre_normalizado = nombre.strip().lower()

while nombre_normalizado != "fin":
    # 1. Verificar si ya existe el producto
    # 2. Si no existe, pedir precio y guardarlo en el diccionario
    # 3. Volver a pedir nombre y normalizar

 Al salir del while:
 Verificar si el diccionario está vacío
 Si no está vacío:
  Recorrer el diccionario para mostrar todos los productos
  Encontrar el producto más caro sin usar valores vacíos especiales
</pre>


```python
productos = {}  # POR QUÉ --> Bucle para agregar y ACUMULAR productos

nombre = input(f"Ingrese el nombre del producto (\"FIN\" Para terminar): ")
nombre_normalizado = nombre.strip().lower()

while nombre_normalizado != "fin":

  # VERIFICAR si existe el producto

  if nombre_normalizado in productos:
    print(f"El producto {nombre_normalizado} ya se encuentra registrado")

  else: #SI NO ESTA REGISTRADO
    # Crear CLAVE -- ASIGNAR valor

    precio = int(input(f"Ingrse el precio del producto: "))

    productos[nombre_normalizado] = precio
    print(f"Producto {nombre_normalizado} agregado al diccionario")

  nombre = input(f"Ingrese el nombre del producto (\"FIN\" Para terminar): ")
  nombre_normalizado = nombre.strip().lower()


# El while es así: 1. Condicion inicial - 2. Validar condicion - 3. ACTUALIZAR condicion

if len(productos) == 0:  #Si esque NO se agregó un producto:
  print(f" No se registró nadototota")

else: #La longitud es distinta de 0

  print(f"\nProductos Registrados")

  for producto, precio in productos.items():  # Clave, Valor, USANDO .items()!!!

    print(f"Producto: {producto} - Precio: {precio}")


# RECORRER EL DICCIONARIO Y BUSCAR EL PRECIO MÁS ALTO

producto_mas_caro = ""   # Utilizar NONE opcional
precio_mas_alto = 0


# REPETICION 0
# PrecioMax = 0 - productoMax = "Ninguno"

# REPETICION 1
# PrecioMax = precio1 - productoMax = producto1

# REPETICION 2
# tengo un precio 2 y si es mayor al primero, es el MÁS ALTO

# REPETICION 3
# tengo precio3 - SI ESTE PRECIO ES MENOR al anterior, entonces precio 3 no es MAIXMO

# REPETICION 4
# tengo precio4, SI ES MAYOR al maximo anterior, el precio 4 es el verdaro MAXIMO

producto_mas_caro = ""   # Utilizar NONE opcional
precio_mas_alto = 0

for producto, precio in productos.items():

  if precio > precio_mas_alto:
    producto_mas_caro = producto
    precio_mas_alto = precio

print(f"Producto más caro: {producto_mas_caro}")
print(f"Precio más alto es: {precio_mas_alto}")
```
