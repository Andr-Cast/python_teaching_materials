<h1>Ayudantía 6: Uso del ciclo <code>while</code> en Python</h1>

<h2>1. ¿Qué es un ciclo <code>while</code>?</h2>
<p>
Un ciclo <code>while</code> permite repetir un bloque de instrucciones
mientras se cumpla una <strong>condición lógica</strong>.
Se usa cuando no sabemos exactamente cuántas veces se repetirá el bloque,
pero sí sabemos cuál es la condición para seguir repitiendo.
</p>

<h2>2. Sintaxis básica</h2>
<pre>
while condición:
    # bloque de instrucciones
    # que se repiten mientras la condición sea verdadera
</pre>

<p>
La condición debe ser una expresión que se pueda evaluar como
<code>True</code> (verdadero) o <code>False</code> (falso).
</p>

<h2>3. Ejemplo 1: Contador simple</h2>
<p>
El siguiente ejemplo muestra un contador que imprime los números del 1 al 5:
</p>

<pre>
contador = 1

while contador &lt;= 5:
    print("Contador:", contador)
    contador = contador + 1  # paso o actualización
</pre>

<ul>
    <li><code>contador = 1</code>: inicializamos la variable.</li>
    <li><code>while contador &lt;= 5:</code>: el ciclo se ejecuta mientras contador sea menor o igual a 5.</li>
    <li>Dentro del <code>while</code> se imprime y se actualiza <code>contador</code>.</li>
    <li>Cuando <code>contador</code> llega a 6, la condición es falsa y el ciclo termina.</li>
</ul>

<h2>4. Importancia de la actualización</h2>
<p>
Dentro del cuerpo del <code>while</code> se debe modificar alguna variable
relacionada con la condición.
Si no se actualiza, la condición puede quedar siempre verdadera
y se produce un <strong>bucle infinito</strong>.
</p>

<h2>5. Ejemplo 2: Suma de números hasta ingresar 0</h2>
<p>
En este ejemplo, el usuario ingresa números y el programa los suma.
El ciclo termina cuando el usuario ingresa 0.
</p>

<pre>
suma = 0
numero = int(input("Ingresa un número (0 para terminar): "))

while numero != 0:
    suma = suma + numero
    numero = int(input("Ingresa un número (0 para terminar): "))

print("La suma total es:", suma)
</pre>

<ul>
    <li>La condición es <code>numero != 0</code>: se repite mientras el número sea distinto de 0.</li>
    <li>Dentro del ciclo se acumula la suma y se vuelve a pedir un número.</li>
    <li>Cuando el usuario ingresa 0, la condición se vuelve falsa y el ciclo termina.</li>
</ul>

<h2>6. Ejemplo 3: Validación de entrada</h2>
<p>
Podemos usar <code>while</code> para obligar al usuario a ingresar un valor válido.
</p>

<pre>
edad = int(input("Ingresa tu edad (valor positivo): "))

while edad &lt;= 0:
    print("Edad no válida. Debe ser un número positivo.")
    edad = int(input("Ingresa tu edad nuevamente: "))

print("Edad registrada:", edad)
</pre>

<ul>
    <li>La condición <code>edad &lt;= 0</code> verifica que la edad no sea válida.</li>
    <li>Mientras la edad sea menor o igual a 0, se sigue pidiendo un nuevo valor.</li>
    <li>Cuando se ingresa una edad positiva, el ciclo termina.</li>
</ul>

<h2>7. Resumen de buenas prácticas con <code>while</code></h2>
<ul>
    <li>Definir claramente la condición del ciclo.</li>
    <li>Inicializar las variables antes del <code>while</code>.</li>
    <li>Asegurarse de actualizar las variables dentro del ciclo.</li>
    <li>Probar casos donde la condición sea verdadera y falsa.</li>
    <li>Evitar bucles infinitos verificando que la condición pueda cambiar a falso.</li>
</ul>

````python
# Ejercicio 5 sin any(), sin break/continue/pass, sin banderas
from random import randint

# --------------------------------------------------------
# Función registrar_compras()
# --------------------------------------------------------
def registrar_compras():
    compras = []
    respuesta = input("Agregar compra? (si/no): ").strip().lower()

    while respuesta == "si":
        producto = input("Producto: ").strip().upper()

        # Validación del precio
        precio = input("Precio: ")
        while not precio.replace(".", "", 1).isdigit():
            print("Precio inválido.")
            precio = input("Precio: ")
        precio = float(precio)

        # Validación de la cantidad
        cantidad = input("Cantidad: ")
        while not cantidad.isdigit():
            print("Cantidad inválida.")
            cantidad = input("Cantidad: ")
        cantidad = int(cantidad)

        compras.append({
            "producto": producto,
            "precio": precio,
            "cantidad": cantidad
        })

        respuesta = input("Agregar compra? (si/no): ").strip().lower()

    return compras


# --------------------------------------------------------
# Función para verificar repetidos sin any()
# --------------------------------------------------------
def nombre_existe(lista, nombre):
    existe = False
    for cliente in lista:
        if cliente["nombre"] == nombre:
            existe = True
    return existe


# --------------------------------------------------------
# Función para registrar un cliente
# --------------------------------------------------------
def registrar_cliente(lista, nombre_limpio):
    # Validar repetido sin any()
    while nombre_existe(lista, nombre_limpio):
        print("El cliente ya existe. Ingresa otro nombre.")
        nuevo = input("Ingresa nombre del cliente: ")
        nombre_limpio = " ".join(nuevo.strip().upper().split())

    # Validación de edad
    edad = input("Ingresa edad: ")
    while not edad.isdigit():
        print("Edad inválida.")
        edad = input("Ingresa edad: ")
    edad = int(edad)

    # Registrar compras
    compras = registrar_compras()

    # Calcular total gastado
    total = 0
    for c in compras:
        total += c["precio"] * c["cantidad"]

    cliente = {
        "id": randint(100000, 999999),
        "nombre": nombre_limpio,
        "edad": edad,
        "compras": compras,
        "total": total
    }

    lista.append(cliente)
    return lista


# --------------------------------------------------------
# Función analizar_clientes()
# --------------------------------------------------------
def analizar_clientes(lista):
    if len(lista) == 0:
        return None, None, 0, []

    total_general = 0

    mayor = lista[0]["total"]
    menor = lista[0]["total"]

    # Encontrar mayor y menor
    for cliente in lista:
        total_general += cliente["total"]

        if cliente["total"] > mayor:
            mayor = cliente["total"]

        if cliente["total"] < menor:
            menor = cliente["total"]

    # Obtener todos los que empatan
    mayores = []
    menores = []

    for cliente in lista:
        if cliente["total"] == mayor:
            mayores.append(cliente)
        if cliente["total"] == menor:
            menores.append(cliente)

    # Eliminar repetidos sin any()
    nombres_unicos = []
    for cliente in lista:
        repetido = False
        for nombre in nombres_unicos:
            if nombre == cliente["nombre"]:
                repetido = True
        if repetido == False:
            nombres_unicos.append(cliente["nombre"])

    return mayores, menores, total_general, nombres_unicos


# --------------------------------------------------------
# PROGRAMA PRINCIPAL (sin banderas)
# --------------------------------------------------------

clientes = []

nombre = input("Ingresa nombre del cliente (SALIR para terminar): ")
nombre = " ".join(nombre.strip().upper().split())

while nombre != "SALIR":
    clientes = registrar_cliente(clientes, nombre)

    nombre = input("Ingresa nombre del cliente (SALIR para terminar): ")
    nombre = " ".join(nombre.strip().upper().split())

# Análisis final
lista_mayores, lista_menores, total, nombres_unicos = analizar_clientes(clientes)

if lista_mayores is None:
    print("No se registraron clientes.")
else:
    print("\nClientes que más gastaron:")
    for c in lista_mayores:
        print(f"{c['nombre']} gastó ${c['total']}")

    print("\nClientes que menos gastaron:")
    for c in lista_menores:
        print(f"{c['nombre']} gastó ${c['total']}")

    print(f"\nGasto total general: ${total}")
    print("Lista de nombres sin repetidos:", nombres_unicos)
````
